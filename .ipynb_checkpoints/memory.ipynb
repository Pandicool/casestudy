```python
def get_completion(prompt , model= "deepseek/deepseek-r1:free"):
    messages = [{"role": "user", "content": prompt}]
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature= 0.7
    )
    return response.choices[0].message.content
    
print(get_completion("what is 1+1?"))
```

```python
from langchain.chains import ConversationChain
from langchain.callbacks.base import BaseCallbackHandler
from langchain.schema import AIMessage, HumanMessage, SystemMessage
```

```python
llm = ChatOpenAI(temperature=0.0, model="deepseek/deepseek-r1:free", api_key= "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", base_url="https://openrouter.ai/api/v1")
```

```python
def count_tokens_from_messages(messages) -> int:
    """Counts tokens from a list of messages."""
    try:
        encoding = tiktoken.encoding_for_model("gpt-3.5-turbo")  # DeepSeek follows OpenAI's ChatML format
    except KeyError:
        encoding = tiktoken.get_encoding("cl100k_base")  # Fallback
    
    text = " ".join([msg.content for msg in messages])  # Convert messages to text
    return len(encoding.encode(text)) 
```

```python
class PatchedConversationTokenBufferMemory(ConversationTokenBufferMemory):
    def get_num_tokens_from_messages(self, messages) -> int:
        return count_tokens_from_messages(messages)
```

```python
memory = PatchedConversationTokenBufferMemory(
    llm=llm,
    max_token_limit=500,  # Adjust as needed
    return_messages=True
)
```

```python
conversation = ConversationChain(llm=llm, memory=memory)
```

```python
user_input = "What is the capital of France?"
response = conversation.predict(input=user_input)
```

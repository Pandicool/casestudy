```python
file= "large_product_list.csv"
```

```python
loader = CSVLoader(file_path=file)
```

```python
docs = loader.load()
```

```python
docs[0]
```

```python
from langchain.embeddings import OpenAIEmbeddings
embedding = OpenAIEmbeddings(openai_api_key="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")
embedding= HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

```

```python
embed = embedding.embed_query("Hi my name is Cool")
```

```python
print(len(embed))
```

```python
print(embed[:1])
```

```python
db = DocArrayInMemorySearch.from_documents(
    docs, 
    embedding
)
```

```python
query = "please select some products which i can wear for friend birthday party and some gift item from my home"
```

```python
docs = db.similarity_search(query)
```

```python
list(docs)
```

```python
len(docs)
```

```python
retriever = db.as_retriever()
```

```python
qdocs = "".join([docs[i].page_content for i in range(len(docs))])
```

```python
response = llm.call_as_llm(f"{qdocs} Question: Please list all your \
wearables  in a table in markdown and summarize each one.") 
```

```python
display(Markdown(response))
```

```python
qa_stuff = RetrievalQA.from_chain_type(
    llm=llm, 
    chain_type="stuff", 
    retriever=retriever, 
    verbose=True
)
```

```python
query =  "Please list all your wearables in a table \
in markdown and summarize each one."
```

```python
response = qa_stuff.run(query)
```

```python
display(Markdown(response))
```

```python
response = index.query(query, llm=llm)
```

```python
from langchain.chains import SimpleSequentialChain
```

```python
llm = ChatOpenAI(temperature=0.9, model="deepseek/deepseek-r1:free", api_key= "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", base_url="https://openrouter.ai/api/v1")
```

```python
prompt_1 = ChatPromptTemplate.from_template(
    "What is the best name to describe \
    a company that makes {product}?"
)
```

```python
prompt_2 = ChatPromptTemplate.from_template(
    "Write a describtion about that {product} in 10 lines with points add some imagnation for future 2050?"
)
```

```python
chain_one = LLMChain(llm=llm, prompt=prompt_1)
```

```python
chain_two = LLMChain(llm=llm, prompt=prompt_2)
```

```python
overall_simple_chain = SimpleSequentialChain(chains=[chain_one, chain_two],
                                             verbose=True
                                            )
```

```python
overall_simple_chain.run(product)
```
